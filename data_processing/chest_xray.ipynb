```python
import os
import pandas as pd
```

```python
train_rows = []

for subdir, _, files in os.walk("../data/chest_xray/train/NORMAL"):
    pre_path = '/'.join(subdir.split('/')[-4:])
    for file in files:
        if file == ".DS_Store":
            continue

        path = os.path.join(subdir, file)
        train_rows.append({'class': 'normal', 'path': pre_path+'/'+file, 'split': 'train'})

for subdir, _, files in os.walk("../data/chest_xray/train/PNEUMONIA"):
    pre_path = '/'.join(subdir.split('/')[-4:])
    for file in files:
        if file == ".DS_Store":
            continue
        
        path = os.path.join(subdir, file)
        train_rows.append({'class': 'pneumonia', 'path': pre_path+'/'+file, 'split': 'train'})
        
train_data = pd.DataFrame(train_rows)
```

```python
train_data.iloc[0].path
```

```python
val_rows  = []

for subdir, _, files in os.walk("../data/chest_xray/val/NORMAL"):
    pre_path = '/'.join(subdir.split('/')[-4:])
    for file in files:
        if file == ".DS_Store":
            continue
        
        path = os.path.join(subdir, file)
        val_rows.append({'class': 'normal', 'path': pre_path+'/'+file, 'split': 'val'})

for subdir, _, files in os.walk("../data/chest_xray/val/PNEUMONIA"):
    pre_path = '/'.join(subdir.split('/')[-4:])
    for file in files:
        if file == ".DS_Store":
            continue
        
        path = os.path.join(subdir, file)
        val_rows.append({'class': 'pneumonia', 'path': pre_path+'/'+file, 'split': 'val'})
        
val_data = pd.DataFrame(val_rows)
```

```python
test_rows  = []

for subdir, _, files in os.walk("../data/chest_xray/test/NORMAL"):
    pre_path = '/'.join(subdir.split('/')[-4:])
    for file in files:
        if file == ".DS_Store":
            continue
        
        path = os.path.join(subdir, file)
        test_rows.append({'class': 'normal', 'path': pre_path+'/'+file, 'split': 'test'})

for subdir, _, files in os.walk("../data/chest_xray/test/PNEUMONIA"):
    pre_path = '/'.join(subdir.split('/')[-4:])
    for file in files:
        if file == ".DS_Store":
            continue
        
        path = os.path.join(subdir, file)
        test_rows.append({'class': 'pneumonia', 'path': pre_path+'/'+file, 'split': 'test'})
        
test_data = pd.DataFrame(test_rows)
```

```python
combined_data = pd.concat([train_data, val_data, test_data])
```

```python
combined_data['class'] = combined_data['class'].map({'pneumonia': 1, 'normal': 0})
```

```python
combined_data.to_csv("../data/chest_xray/chest_xray_dataset.csv")
```
