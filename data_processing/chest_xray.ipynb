```python
import os
import pandas as pd
```

```python
train_rows = []

for subdir, _, files in os.walk("/home/ara/Documents/projects/deep_learning_project/data/chest_xray/train/NORMAL"):
    pre_path = '/'.join(subdir.split('/')[-4:])
    for file in files:
        path = os.path.join(subdir, file)
        train_rows.append({'class': 'normal', 'path': pre_path+file, 'split': 'train'})

for subdir, _, files in os.walk("/home/ara/Documents/projects/deep_learning_project/data/chest_xray/train/PNEUMONIA"):
    pre_path = '/'.join(subdir.split('/')[-4:])
    for file in files:
        path = os.path.join(subdir, file)
        train_rows.append({'class': 'pneumonia', 'path': pre_path+file, 'split': 'train'})
        
train_data = pd.DataFrame(train_rows)
# train_data.to_csv("chest_xray_train.csv")
```
